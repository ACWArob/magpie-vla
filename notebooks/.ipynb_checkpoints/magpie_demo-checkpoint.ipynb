````python
import subprocess, os, time, signal

# ── Edit CAMERA_CMD if your launch command differs ──────────────────
CAMERA_CMD = 'ros2 run realsense2_camera realsense2_camera_node --ros-args -r __ns:=/camera/gripper_camera'
SAM3_PYTHON = os.path.expanduser('~/sam3_env/bin/python3')
SAM3_SCRIPT = os.path.expanduser('~/magpie_control/scripts/sam3_infer.py')
ROS_SOURCE  = 'source /opt/ros/humble/setup.bash && source ~/ws_ctrl/install/setup.bash'

def _ros(cmd, log):
    f = open(log, 'w')
    return subprocess.Popen(f'{ROS_SOURCE} && {cmd}', shell=True,
                            executable='/bin/bash', stdout=f, stderr=subprocess.STDOUT)

if 'PROCS' not in globals():
    PROCS = {}

def launch_all():
    jobs = {
        'ur5_node':     ('ros2 run magpie_control ur5_node',     '/tmp/ur5_node.log'),
        'gripper_node': ('ros2 run magpie_control gripper_node', '/tmp/gripper_node.log'),
        'camera':       (CAMERA_CMD,                             '/tmp/camera_node.log'),
        'sam3':         (f'{SAM3_PYTHON} {SAM3_SCRIPT}',         '/tmp/sam3_server.log'),
    }
    for name, (cmd, log) in jobs.items():
        if name in PROCS and PROCS[name].poll() is None:
            print(f'  {name:<14} already running (pid {PROCS[name].pid})')
            continue
        PROCS[name] = _ros(cmd, log)
        print(f'  {name:<14} started  pid={PROCS[name].pid}  log={log}')

launch_all()
print('Waiting 10s for nodes to initialise...')
time.sleep(10)

print('\nStatus:')
for name, p in PROCS.items():
    state = 'RUNNING' if p.poll() is None else f'EXITED (code {p.poll()})'
    print(f'  {name:<14} {state}')
    if p.poll() is not None:
        log = f'/tmp/{name.replace("_","")}.log' if name != 'sam3' else '/tmp/sam3_server.log'
        try:
            tail = open(f'/tmp/{name}.log').read()[-400:]
            print(f'    last log: {tail}')
        except: pass
print('\nReady. Run the cells below.')
````

````python
import sys, os, time, re, json, tempfile, base64, socket as _sock, subprocess
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from mpl_toolkits.mplot3d import Axes3D
from IPython.display import display
import cv2

%matplotlib inline
plt.rcParams['figure.figsize'] = (13, 5)
plt.rcParams['font.size'] = 11

sys.path.insert(0, os.path.expanduser('~/magpie_control/scripts'))
sys.path.insert(0, os.path.expanduser('~/magpie_control/src'))

# ── Config ──────────────────────────────────────────────────────────
OBJECT         = 'red block'   # set None to auto-detect via Gemini
SAM3_SOCK      = '/tmp/sam3.sock'
SAM3_SCRIPT    = os.path.expanduser('~/magpie_control/scripts/sam3_infer.py')
SAM3_PYTHON    = os.path.expanduser('~/sam3_env/bin/python3')
GEMINI_API_KEY = os.environ.get('GEMINI_API_KEY', '')
GRIPPER_LENGTH = 0.231   # m — wrist flange to fingertip
HARD_FLOOR_Z   = 0.030   # m — physical table surface
APPROACH_H     = 0.10    # m — approach height above object

if not GEMINI_API_KEY:
    print('[WARN] GEMINI_API_KEY not set')
print('Config OK  |  OBJECT =', repr(OBJECT))
````

````python
import rclpy
from rclpy.node import Node
from sensor_msgs.msg import Image as RosImage, CameraInfo
from geometry_msgs.msg import PoseStamped, Pose
from std_srvs.srv import Trigger
from cv_bridge import CvBridge
from magpie_msgs.msg import GripperState
from magpie_msgs.srv import MoveLinear, SetGripperForce, SetGripperPosition
from magpie_control import poses
from magpie_control.homog_utils import homog_xform, R_krot
from magpie_control.gripper_arc import fingertip_drop as _gripper_drop
from google import genai
from google.genai import types as gtypes
from pointcloud_utils import build_segmented_pcd, denoise_pcd, analyse_pcd

_TCP_TO_CAM = homog_xform(R_krot([0,0,1], -np.pi/2), [0,0,0.120])

def _quat_to_rv(w, x, y, z):
    angle = 2.0 * np.arccos(np.clip(w, -1.0, 1.0))
    s = np.sin(angle / 2.0)
    if s < 1e-10: return np.zeros(3)
    return angle * np.array([x, y, z]) / s

def _rv_to_quat(rv):
    angle = np.linalg.norm(rv)
    if angle < 1e-10: return (1.0, 0.0, 0.0, 0.0)
    ax = rv / angle; s = np.sin(angle/2.0)
    return (np.cos(angle/2.0), ax[0]*s, ax[1]*s, ax[2]*s)

def _mat_to_pose_msg(mat):
    vec = poses.pose_mtrx_to_vec(np.array(mat))
    w, x, y, z = _rv_to_quat(np.array(vec[3:]))
    msg = Pose()
    msg.position.x, msg.position.y, msg.position.z = vec[0], vec[1], vec[2]
    msg.orientation.w, msg.orientation.x = w, x
    msg.orientation.y, msg.orientation.z = y, z
    return msg

def _straight_down_rot(deg=0.0):
    t = np.radians(deg); c, s = np.cos(t), np.sin(t)
    return np.array([[c,s,0],[s,-c,0],[0,0,-1]])

print('ROS2 imports OK')
````

````python
class DemoNode(Node):
    def __init__(self):
        super().__init__('magpie_demo')
        self.bridge = CvBridge()
        self.color = None; self.depth = None
        self.caminfo = None; self.tcp = None; self.gs = None

        CAM = '/camera/gripper_camera'
        self.create_subscription(RosImage,     CAM+'/color/image_raw',      self._c_cb, 1)
        self.create_subscription(RosImage,     CAM+'/depth/image_rect_raw', self._d_cb, 1)
        self.create_subscription(CameraInfo,   CAM+'/color/camera_info',    self._i_cb, 1)
        self.create_subscription(PoseStamped,  '/arm/tcp_pose',             self._t_cb, 1)
        self.create_subscription(GripperState, '/gripper/state',            self._g_cb, 1)

        self.cli_move  = self.create_client(MoveLinear,        '/arm/move_l')
        self.cli_teach = self.create_client(Trigger,            '/arm/teach_mode')
        self.cli_open  = self.create_client(Trigger,            '/gripper/open')
        self.cli_close = self.create_client(Trigger,            '/gripper/close')
        self.cli_force = self.create_client(SetGripperForce,    '/gripper/set_force')
        self.cli_pos   = self.create_client(SetGripperPosition, '/gripper/set_position')

    def _c_cb(self, m): self.color   = self.bridge.imgmsg_to_cv2(m, 'rgb8')
    def _d_cb(self, m): self.depth   = self.bridge.imgmsg_to_cv2(m, 'passthrough')
    def _i_cb(self, m): self.caminfo = m
    def _g_cb(self, m): self.gs = m
    def _t_cb(self, m):
        p = m.pose
        rv = _quat_to_rv(p.orientation.w, p.orientation.x, p.orientation.y, p.orientation.z)
        self.tcp = poses.pose_vec_to_mtrx([p.position.x, p.position.y, p.position.z, *rv])

    def spin(self, n=10, t=0.15):
        for _ in range(n): rclpy.spin_once(self, timeout_sec=t)

    def _call(self, cli, req, timeout=30.0):
        if not cli.wait_for_service(timeout_sec=3.0):
            raise RuntimeError(f'Service unavailable: {cli.srv_name}')
        fut = cli.call_async(req)
        rclpy.spin_until_future_complete(self, fut, timeout_sec=timeout)
        return fut.result()

    def move_l(self, mat, speed=0.10, accel=0.2):
        req = MoveLinear.Request()
        req.target_pose = _mat_to_pose_msg(mat)
        req.speed = speed; req.acceleration = accel; req.async_mode = False
        resp = self._call(self.cli_move, req)
        if not resp.success: raise RuntimeError(resp.message)

    def open_g(self):  return self._call(self.cli_open,  Trigger.Request())
    def close_g(self): return self._call(self.cli_close, Trigger.Request())

    def set_force(self, n):
        req = SetGripperForce.Request(); req.max_force = float(n)
        return self._call(self.cli_force, req)

    def set_pos(self, mm):
        req = SetGripperPosition.Request(); req.position = float(max(0.0, mm))
        return self._call(self.cli_pos, req)

    def disable_teach(self):
        if not self.cli_teach.wait_for_service(timeout_sec=2.0): return
        r = self._call(self.cli_teach, Trigger.Request())
        if 'enabled' in r.message.lower():
            self._call(self.cli_teach, Trigger.Request())

    def wait_data(self, timeout=12.0):
        t0 = time.time()
        while time.time()-t0 < timeout:
            rclpy.spin_once(self, timeout_sec=0.15)
            if all(v is not None for v in [self.color,self.depth,self.caminfo,self.tcp]):
                return True
        return False

try:
    rclpy.init()
except RuntimeError:
    pass

node = DemoNode()
ok   = node.wait_data(12.0)
print('Data ready:', ok,
      ' | color:', node.color is not None,
      'depth:', node.depth is not None,
      'tcp:', node.tcp is not None)
````

## 4. Object Detection
Three methods shown side-by-side. Also demonstrates **specified object** (`OBJECT`) vs **auto-detect** (Gemini identifies it from the image).

### 4a. ViT — OWL-ViT Zero-Shot Open-Vocabulary Detection
`google/owlvit-base-patch32` — takes text query + image, returns bounding boxes with confidence scores. Runs fully locally, no API key needed.

````python
try:
    import torch
    from PIL import Image as PILImage
    from transformers import OwlViTProcessor, OwlViTForObjectDetection

    t0 = time.time()
    processor = OwlViTProcessor.from_pretrained('google/owlvit-base-patch32')
    model     = OwlViTForObjectDetection.from_pretrained('google/owlvit-base-patch32')
    model.eval()
    print(f'Model loaded in {time.time()-t0:.1f}s')

    pil_img = PILImage.fromarray(color)
    query   = OBJECT if OBJECT else 'graspable object'
    texts   = [[query, f'a {query}', f'{query} on a table']]
    inputs  = processor(text=texts, images=pil_img, return_tensors='pt')

    t1 = time.time()
    with torch.no_grad():
        outputs = model(**inputs)
    t_inf = time.time() - t1

    target_sizes = torch.tensor([pil_img.size[::-1]])
    results = processor.post_process_object_detection(
        outputs, threshold=0.05, target_sizes=target_sizes)[0]

    vit_boxes  = results['boxes'].numpy()
    vit_scores = results['scores'].numpy()

    print(f'Inference: {t_inf*1000:.0f} ms  |  Detections: {len(vit_boxes)}')
    for i, (b, s) in enumerate(zip(vit_boxes[:5], vit_scores[:5])):
        print(f'  [{i}] score={s:.3f}  box={b.astype(int).tolist()}')

    fig, ax = plt.subplots(figsize=(10, 6))
    ax.imshow(color)
    cmap = plt.cm.YlGn
    for box, score in zip(vit_boxes, vit_scores):
        x1,y1,x2,y2 = box
        ec = cmap(float(score))
        ax.add_patch(patches.Rectangle((x1,y1),x2-x1,y2-y1,
                     linewidth=2,edgecolor=ec,facecolor='none'))
        ax.text(x1,y1-6,f'{score:.2f}',color=ec,fontsize=9,fontweight='bold')
    ax.set_title(f'OWL-ViT  query="{query}"  {len(vit_boxes)} detections  ({t_inf*1000:.0f} ms)')
    ax.axis('off'); plt.tight_layout(); plt.show()

except ImportError as e:
    print(f'[SKIP] transformers not installed: {e}')
    print('  pip install transformers torch pillow')
    vit_boxes, vit_scores = np.zeros((0,4)), np.array([])
````

### 4b. VLM — Gemini 2.5-Flash
Two modes:
- **Specified object**: asks Gemini to locate `OBJECT` in the image
- **Auto-detect** (`OBJECT = None`): asks Gemini to identify the main graspable object — the output becomes the query for SAM3 downstream

````python
_, buf = cv2.imencode('.jpg', cv2.cvtColor(color, cv2.COLOR_RGB2BGR))
gclient = genai.Client(api_key=GEMINI_API_KEY)

# ── Auto-detect mode ────────────────────────────────────────────────
resp_auto = gclient.models.generate_content(
    model='gemini-2.5-flash',
    contents=[
        gtypes.Part.from_bytes(data=buf.tobytes(), mime_type='image/jpeg'),
        ('What is the main graspable object in this image? '
         'Reply with only the object name, 2-4 words maximum, no punctuation.'),
    ],
)
raw_auto = resp_auto.text.strip()
m = re.search(r'[.!?\n]\s*([^\n.!?]{2,40})\s*$', raw_auto)
auto_object = (m.group(1).strip() if m else raw_auto.split('\n')[-1].strip()).lower()
auto_object = ' '.join(auto_object.split()[:4])
print(f'Auto-detect result: "{auto_object}"')

# ── Specified object mode ────────────────────────────────────────────
active_object = OBJECT if OBJECT else auto_object

resp_spec = gclient.models.generate_content(
    model='gemini-2.5-flash',
    contents=[
        gtypes.Part.from_bytes(data=buf.tobytes(), mime_type='image/jpeg'),
        (f'Locate the "{active_object}" in this image. '
         f'Describe: (1) position in image (top/mid/bot, left/center/right), '
         f'(2) approximate pixel bounding box as x1,y1,x2,y2, '
         f'(3) suitability for robot grasping (yes/no + reason). '
         f'Be concise.'),
    ],
)
lines = [l for l in resp_spec.text.strip().split('\n') if l.strip()][:8]
print(f'\nSpecified mode ("{active_object}"):')
for l in lines: print(' ', l)

# ── Side-by-side visualisation ───────────────────────────────────────
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
ax1.imshow(color)
ax1.text(10, 25, f'Auto-detected:\n"{auto_object}"', color='white', fontsize=11,
         bbox=dict(boxstyle='round,pad=0.4', fc='black', alpha=0.75))
ax1.set_title('VLM Auto-Detect (OBJECT=None)'); ax1.axis('off')

ax2.imshow(color)
desc = '\n'.join(lines[:4])
ax2.text(10, 25, desc, color='white', fontsize=9,
         bbox=dict(boxstyle='round,pad=0.4', fc='navy', alpha=0.75))
ax2.set_title(f'VLM Specified ("{active_object}")'); ax2.axis('off')

plt.suptitle('Gemini 2.5-Flash — Auto-Detect vs Specified Object', fontsize=12)
plt.tight_layout(); plt.show()

# Pass active_object downstream to SAM3 / point cloud sections
ACTIVE_OBJECT = active_object
print(f'\nUsing "{ACTIVE_OBJECT}" for downstream detection.')
````

### 4c. SAM3 — Open-Vocabulary Segmentation
Two modes:
- **Socket mode** (fast): server already running, Unix socket `/tmp/sam3.sock`, ~1s round-trip
- **Cold start** (slow): launches `sam3_infer.py` subprocess, ~30-60s first load, useful when no server is running

````python
def sam3_socket(image_rgb, query, sock_path=SAM3_SOCK):
    tmp = tempfile.mktemp(suffix='.jpg')
    cv2.imwrite(tmp, cv2.cvtColor(image_rgb, cv2.COLOR_RGB2BGR))
    try:
        with _sock.socket(_sock.AF_UNIX, _sock.SOCK_STREAM) as s:
            s.connect(sock_path)
            s.sendall((json.dumps({'image': tmp, 'query': query}) + '\n').encode())
            resp = b''
            while True:
                chunk = s.recv(65536)
                if not chunk: break
                resp += chunk
        data = json.loads(resp.decode().strip())
        if 'error' in data: raise RuntimeError(data['error'])
        boxes  = np.array(data['boxes'],  dtype=float)
        scores = np.array(data['scores'], dtype=float)
        mask = None
        if data.get('mask_b64') and len(boxes) > 0:
            raw  = base64.b64decode(data['mask_b64'])
            h, w = data['mask_shape']
            mask = np.frombuffer(raw, dtype=np.uint8).reshape(h,w).astype(bool)
        return boxes, scores, mask, data
    finally:
        if os.path.exists(tmp): os.unlink(tmp)

# ── Socket mode ──────────────────────────────────────────────────────
print('=== SAM3 SOCKET MODE ===')
t0 = time.time()
try:
    s_boxes, s_scores, s_mask, _ = sam3_socket(color, ACTIVE_OBJECT)
    t_sock = time.time() - t0
    best_s = int(np.argmax(s_scores)) if len(s_scores) > 0 else 0
    print(f'  Time: {t_sock:.2f}s  |  Detections: {len(s_boxes)}  '
          f'|  Best score: {s_scores[best_s]:.3f}  '
          f'|  Mask px: {s_mask.sum() if s_mask is not None else 0}')
    socket_ok = True
except Exception as e:
    print(f'  [FAIL] {e}  — is sam3_infer.py running?')
    s_boxes, s_scores, s_mask = np.zeros((0,4)), np.array([]), None
    t_sock = 0.0; socket_ok = False

# ── Cold start mode ──────────────────────────────────────────────────
print('\n=== SAM3 COLD START MODE ===')
COLD_SOCK = '/tmp/sam3_cold.sock'

t_cold_start = time.time()
cold_proc = subprocess.Popen(
    [SAM3_PYTHON, SAM3_SCRIPT, '--sock', COLD_SOCK],
    stdout=subprocess.PIPE, stderr=subprocess.PIPE
)
print(f'  Subprocess PID {cold_proc.pid} — waiting for model load (~30-60s)...')

loaded = False
for _ in range(90):            # wait up to 90s
    time.sleep(1.0)
    if os.path.exists(COLD_SOCK):
        loaded = True; break
    if cold_proc.poll() is not None:
        print('  [FAIL] Subprocess exited early')
        break

t_load = time.time() - t_cold_start
if loaded:
    print(f'  Model loaded in {t_load:.1f}s')
    t1 = time.time()
    try:
        c_boxes, c_scores, c_mask, _ = sam3_socket(color, ACTIVE_OBJECT, COLD_SOCK)
        t_cold_infer = time.time() - t1
        best_c = int(np.argmax(c_scores)) if len(c_scores) > 0 else 0
        print(f'  Inference: {t_cold_infer:.2f}s  |  Detections: {len(c_boxes)}  '
              f'|  Best score: {c_scores[best_c]:.3f}  '
              f'|  Mask px: {c_mask.sum() if c_mask is not None else 0}')
        cold_ok = True
    except Exception as e:
        print(f'  [FAIL] {e}'); c_boxes,c_scores,c_mask = np.zeros((0,4)),np.array([]),None
        t_cold_infer = 0.0; cold_ok = False
else:
    print(f'  [FAIL] Sock not found after {t_load:.0f}s')
    c_boxes,c_scores,c_mask = np.zeros((0,4)),np.array([]),None
    t_cold_infer = 0.0; cold_ok = False

cold_proc.terminate()
if os.path.exists(COLD_SOCK): os.unlink(COLD_SOCK)

# Keep best result for downstream cells
best_boxes  = s_boxes  if socket_ok else c_boxes
best_scores = s_scores if socket_ok else c_scores
best_mask   = s_mask   if socket_ok else c_mask
print(f'\nUsing {"socket" if socket_ok else "cold"} result downstream.')
````

````python
# ── SAM3 visual comparison ───────────────────────────────────────────
def draw_det(ax, img, boxes, scores, mask, title):
    vis = img.copy()
    if mask is not None:
        vis[mask] = (vis[mask]*0.45 + np.array([0,200,0])*0.55).astype(np.uint8)
    ax.imshow(vis)
    if len(boxes) > 0:
        best = int(np.argmax(scores))
        x1,y1,x2,y2 = boxes[best].astype(int)
        ax.add_patch(patches.Rectangle((x1,y1),x2-x1,y2-y1,
                     linewidth=3, edgecolor='lime', facecolor='none'))
        ax.text(x1, y1-8, f'score={scores[best]:.3f}',
                color='lime', fontsize=10, fontweight='bold')
    ax.set_title(title, fontsize=11); ax.axis('off')

fig, axes = plt.subplots(1, 3, figsize=(17, 5))
axes[0].imshow(color); axes[0].set_title('Original'); axes[0].axis('off')

sock_label = (f'Socket mode\n{t_sock:.2f}s  score={s_scores[int(np.argmax(s_scores))]:.3f}'
              if socket_ok and len(s_scores)>0 else 'Socket mode\nFAILED')
cold_label = (f'Cold start\nload {t_load:.0f}s + infer {t_cold_infer:.2f}s'
              if cold_ok and len(c_scores)>0 else 'Cold start\nFAILED / not run')

draw_det(axes[1], color, s_boxes, s_scores, s_mask, sock_label)
draw_det(axes[2], color, c_boxes, c_scores, c_mask, cold_label)

plt.suptitle(f'SAM3 — Socket vs Cold Start  |  query: "{ACTIVE_OBJECT}"', fontsize=12)
plt.tight_layout(); plt.show()

# Timing bar chart
fig2, ax = plt.subplots(figsize=(7, 3))
labels = ['Socket\nround-trip', 'Cold start\nmodel load', 'Cold start\ninference']
vals   = [t_sock, t_load, t_cold_infer]
clrs   = ['steelblue', 'tomato', 'orange']
ax.barh(labels, vals, color=clrs)
for i, v in enumerate(vals):
    ax.text(v+0.3, i, f'{v:.1f}s', va='center')
ax.set_xlabel('seconds'); ax.set_title('SAM3 Latency: Socket vs Cold Start')
plt.tight_layout(); plt.show()
````

## 6. 6D Pose / 3D Centroid
Back-project mask centroid pixel through camera intrinsics + TCP→world transform to get world-frame (x, y, z).

````python
if best_mask is not None and best_mask.any():
    ys, xs = np.where(best_mask)
    u, v   = int(xs.mean()), int(ys.mean())
    valid  = depth_s[best_mask].astype(float)
    valid  = valid[valid > 0]
else:
    h, w = depth_s.shape; u, v = w//2, h//2
    valid = np.array([])

depth_m  = float(np.median(valid)) / 1000.0 if len(valid) > 0 else 0.0
fx, fy   = ci.k[0], ci.k[4]
cx, cy   = ci.k[2], ci.k[5]
p_cam    = np.array([(u-cx)*depth_m/fx, (v-cy)*depth_m/fy, depth_m])
T_world  = tcp_s @ _TCP_TO_CAM
p_world  = (T_world @ np.array([*p_cam, 1.0]))[:3]

print(f'Centroid pixel : ({u}, {v})')
print(f'Depth          : {depth_m*1000:.1f} mm  ({len(valid)} valid pixels in mask)')
print(f'Camera frame   : x={p_cam[0]:.4f}  y={p_cam[1]:.4f}  z={p_cam[2]:.4f} m')
print(f'World frame    : x={p_world[0]:.4f}  y={p_world[1]:.4f}  z={p_world[2]:.4f} m')
print(f'\nTCP->World 4x4 (T = tcp @ _TCP_TO_CAM):')
print(np.round(T_world, 4))

fig, ax = plt.subplots(figsize=(9, 6))
vis2 = color_s.copy()
if best_mask is not None:
    vis2[best_mask] = (vis2[best_mask]*0.45 + np.array([0,150,255])*0.55).astype(np.uint8)
ax.imshow(vis2)
ax.plot(u, v, 'r+', markersize=22, markeredgewidth=3)
ax.text(u+12, v, f'({p_world[0]:.3f}, {p_world[1]:.3f}, {p_world[2]:.3f}) m\nworld frame',
        color='white', fontsize=10,
        bbox=dict(boxstyle='round,pad=0.3', fc='black', alpha=0.7))
ax.set_title(f'3D Centroid  pixel=({u},{v})  depth={depth_m*1000:.0f}mm')
ax.axis('off'); plt.tight_layout(); plt.show()
````

## 7. PCA + Gripper Alignment (Super Side Quest)
`analyse_pcd()` decomposes the point cloud into principal axes. The major axis gives the grasp angle — gripper fingers are rotated perpendicular to it.

````python
pca      = analyse_pcd(pts_cln)
centroid = pca['centroid']
extent_m = pca['extent_m']
angle    = pca['grasp_angle_deg']
offset   = float(np.clip(extent_m[2]/4.0, 0.005, 0.015))

print(f'PCA centroid : ({centroid[0]:.4f}, {centroid[1]:.4f}, {centroid[2]:.4f}) m')
print(f'extent_m     : major={extent_m[0]*1000:.1f}mm  minor={extent_m[1]*1000:.1f}mm  normal={extent_m[2]*1000:.1f}mm')
print(f'grasp_angle  : {angle:.1f} deg  (gripper fingers perpendicular to major axis)')
print(f'auto_offset  : {offset*1000:.0f} mm  (= normal/4, clamped [5, 15] mm)')

# Compute eigenvectors for plotting
pts_c = pts_cln - centroid
cov   = np.cov(pts_c.T)
evals, evecs = np.linalg.eigh(cov)
idx   = np.argsort(evals)[::-1]
evecs = evecs[:, idx]; evals = evals[idx]
scale = np.sqrt(np.abs(evals))

fig = plt.figure(figsize=(14, 6))
ax  = fig.add_subplot(121, projection='3d')
ax.scatter(pts_cln[:,0], pts_cln[:,1], pts_cln[:,2],
           c=pts_cln[:,2], cmap='plasma', s=4, alpha=0.45)
axis_labels = ['Major', 'Minor', 'Normal (height)']
axis_colors = ['red', 'green', 'blue']
for i in range(3):
    v = evecs[:,i] * scale[i]
    ax.quiver(*centroid, *v, color=axis_colors[i],
              label=f'{axis_labels[i]} {scale[i]*1000:.0f}mm', linewidth=2.5)
ax.scatter(*centroid, color='yellow', s=120, zorder=5)
ax.set_xlabel('X'); ax.set_ylabel('Y'); ax.set_zlabel('Z')
ax.set_title(f'PCA axes  grasp_angle={angle:.1f}\u00b0')
ax.legend(fontsize=9, loc='upper left')

# Show rotation in XY plane
ax2 = fig.add_subplot(122)
ax2.scatter(pts_cln[:,0], pts_cln[:,1], c=pts_cln[:,2], cmap='plasma', s=5, alpha=0.5)
theta = np.radians(angle)
L = scale[0]
ax2.annotate('', xy=(centroid[0]+L*np.cos(theta), centroid[1]+L*np.sin(theta)),
             xytext=(centroid[0]-L*np.cos(theta), centroid[1]-L*np.sin(theta)),
             arrowprops=dict(arrowstyle='<->', color='red', lw=2.5))
ax2.annotate('', xy=(centroid[0]+L*np.cos(theta+np.pi/2), centroid[1]+L*np.sin(theta+np.pi/2)),
             xytext=centroid[:2],
             arrowprops=dict(arrowstyle='->', color='lime', lw=2.0))
ax2.text(centroid[0]+L*np.cos(theta+np.pi/2)+0.005, centroid[1]+L*np.sin(theta+np.pi/2),
         'gripper\napproach', fontsize=9, color='lime')
ax2.set_aspect('equal'); ax2.set_xlabel('X (m)'); ax2.set_ylabel('Y (m)')
ax2.set_title(f'Top-down: major axis (red) + gripper direction (green)')
plt.suptitle('PCA + Gripper Alignment', fontsize=12)
plt.tight_layout(); plt.show()
````

## 8. Verify: Move Arm to Computed Pose
Build approach pose from PCA centroid + grasp angle. Run safety check. Move arm to approach height only (does **not** descend to grasp).

````python
R_grasp  = _straight_down_rot(angle)
approach = np.eye(4); approach[:3,:3] = R_grasp
approach[:3,3] = [p_world[0], p_world[1],
                  p_world[2] + APPROACH_H + GRIPPER_LENGTH]

grasp_pose = approach.copy()
grasp_pose[2,3] = p_world[2] - offset + GRIPPER_LENGTH

approach_fz = p_world[2] + APPROACH_H
grasp_fz    = p_world[2] - offset
drop        = _gripper_drop(103.6, 0.0)   # worst-case (open fingers)
grasp_eff_z = grasp_fz - drop

ok_approach = approach_fz  >= HARD_FLOOR_Z
ok_grasp    = grasp_eff_z  >= HARD_FLOOR_Z

print('=== SAFETY CHECK ===')
print(f'  Object centroid z : {p_world[2]:.4f} m')
print(f'  Approach fingertip: {approach_fz:.4f} m  {"OK" if ok_approach else "BELOW FLOOR"}')
print(f'  Grasp open fing.  : {grasp_fz:.4f} m')
print(f'  Arc drop (worst)  : {drop*1000:.0f} mm')
print(f'  Grasp eff. z      : {grasp_eff_z:.4f} m  {"OK" if ok_grasp else "WOULD HIT FLOOR"}')
print(f'  Floor limit       : {HARD_FLOOR_Z:.3f} m')
print(f'  SAFE TO EXECUTE   : {"YES" if ok_approach and ok_grasp else "NO"}')

if ok_approach and ok_grasp:
    print(f'\nMoving to approach pose (TCP z={approach[2,3]:.3f} m)...')
    node.disable_teach()
    node.move_l(approach, speed=0.08)
    time.sleep(0.5); node.spin(6)
    cur = poses.pose_mtrx_to_vec(node.tcp)
    print(f'Arm at: x={cur[0]:.3f}  y={cur[1]:.3f}  z={cur[2]:.3f} m  (fingertip ~{approach_fz:.3f} m)')
    print('Returning home...')
    node.move_l(tcp_home, speed=0.08)
else:
    print('SKIPPED — safety check failed.')
````

## 9. Gripper Tool Length Accounting
Why `GRIPPER_LENGTH = 0.231 m` is critical: the UR5 controls the wrist flange, not the fingertip. Without accounting for it, commanded z and actual fingertip z are 231mm apart — guaranteed crash.

````python
print('Gripper tool length accounting:')
print(f'  Object centroid z  = {p_world[2]:.4f} m  (top surface from depth camera)')
print(f'  Approach height    = {APPROACH_H:.3f} m  (above centroid)')
print(f'  GRIPPER_LENGTH     = {GRIPPER_LENGTH:.3f} m  (flange -> fingertip)')
print(f'  TCP z commanded    = {approach[2,3]:.4f} m  (= obj_z + approach_h + tool_len)')
print(f'  Fingertip z actual = {approach_fz:.4f} m  (= TCP z - tool_len)')
print()
wrong_tcp_z = p_world[2] + APPROACH_H   # what you'd command WITHOUT tool length
print(f'  WITHOUT tool length: TCP z = {wrong_tcp_z:.4f} m')
print(f'    -> fingertip would be at TCP z (no offset) = {wrong_tcp_z:.4f} m')
print(f'    -> but TCP is 0.231m ABOVE fingertip')
print(f'    -> so fingertip = {wrong_tcp_z:.4f} - 0.231 = {wrong_tcp_z-GRIPPER_LENGTH:.4f} m')
print(f'    -> which is {(p_world[2]-offset - (wrong_tcp_z-GRIPPER_LENGTH))*1000:.0f} mm below intended grasp z -> CRASH')

fig, ax = plt.subplots(figsize=(7, 7))
items = [
    ('TCP flange (commanded)',              approach[2,3],      'royalblue'),
    (f'Fingertip (actual = TCP-{GRIPPER_LENGTH}m)', approach_fz, 'limegreen'),
    ('Object centroid (top surface)',        p_world[2],        'orange'),
    ('Intended grasp fingertip z',           grasp_fz,          'gold'),
    ('Table floor (HARD_FLOOR_Z)',           HARD_FLOOR_Z,      'saddlebrown'),
]
for i, (lbl, z, c) in enumerate(items):
    ax.barh(i, z, color=c, alpha=0.85)
    ax.text(z+0.002, i, f'{z:.3f} m', va='center', fontsize=10)
ax.set_yticks(range(len(items)))
ax.set_yticklabels([it[0] for it in items], fontsize=10)
ax.axvline(HARD_FLOOR_Z, color='red', lw=1.5, ls='--', label='Floor limit')
ax.set_xlabel('Height (m)'); ax.set_title('Gripper Tool Length — Approach Pose')
ax.legend(); plt.tight_layout(); plt.show()
````

## 10. DeliGrasp Parameters (Gemini)
Gemini generates physics-grounded gripper parameters: mass, friction, stiffness, aperture, and force schedule.

````python
DG_PROMPT = (
    'Control a robot gripper with force control and contact information. '
    'The gripper has a measurable max force of 16N and min force of 0.15N, '
    'a maximum aperture of 105mm and a minimum aperture of 1mm.\n\n'
    'Describe the grasp strategy using the following form:\n\n'
    '[start of description]\n'
    '* This is a new grasp.\n'
    '* In accordance with the user instruction, this grasp should be [GRASP_DESCRIPTION: <str>].\n'
    '* This is a complete grasp.\n'
    '* This grasp does not contain multiple grasps.\n'
    '* This grasp is for an object with {CHOICE: [high, medium, low]} weight.\n'
    '* The object has an approximate mass of [PNUM: 0.0] grams\n'
    '* This grasp is for an object with {CHOICE: [high, medium, low]} compliance.\n'
    '* The object has an approximate spring constant of [PNUM: 0.0] Newtons per meter.\n'
    '* The gripper and object have an approximate friction coefficient of [PNUM: 0.0]\n'
    '* This grasp should set the goal aperture to [PNUM: 0.0] mm.\n'
    '* If the gripper slips, this grasp should close an additional [PNUM: 0.0] mm.\n'
    '* If the gripper slips, this grasp should increase the output force by [PNUM: 0.0] Newtons.\n'
    '[end of description]\n\n'
    'Rules: Replace {CHOICE:...} with one choice. Replace [PNUM: 0.0] with a positive number. '
    'Always start with [start of description] and end with [end of description].'
)

def call_descriptor(obj_name):
    client = genai.Client(api_key=GEMINI_API_KEY)
    resp = client.models.generate_content(
        model='gemini-2.5-flash',
        contents=f'Pick up the {obj_name}.',
        config=gtypes.GenerateContentConfig(system_instruction=DG_PROMPT),
    )
    return resp.text

def parse_descriptor(text):
    m = re.search(r'\[start of description\](.*?)\[end of description\]',
                  text, re.DOTALL | re.IGNORECASE)
    if not m: raise ValueError('Tags not found')
    body = m.group(1)
    NUM  = r'([0-9]+(?:\.[0-9]+)?)'
    def g(pat):
        hit = re.search(pat, body, re.IGNORECASE)
        return float(hit.group(1)) if hit else None
    mass_g   = g(rf'approximate mass of {NUM} grams')
    k        = g(rf'spring constant of {NUM} Newtons per meter')
    mu       = g(rf'friction coefficient of {NUM}')
    aperture = g(rf'goal aperture to {NUM} mm')
    closure  = g(rf'close an additional {NUM} mm')
    add_f    = g(rf'increase the output force by {NUM} Newtons')
    if None in [mass_g, k, mu, aperture, closure]:
        raise ValueError(f'Parse failed mass={mass_g} k={k} mu={mu} ap={aperture}')
    init_f = min(16.0, max(0.15, (mass_g/1000*9.81)/max(mu, 1e-6)))
    add_force = add_f if add_f else max(0.05, closure*k*0.0001)
    return dict(mass_g=mass_g, k=k, mu=mu, aperture_mm=aperture,
                closure_mm=closure, initial_force=init_f, add_force=add_force)

raw = call_descriptor(ACTIVE_OBJECT)
print(raw)
gp  = parse_descriptor(raw)

print('\n=== Parsed Parameters ===')
rows = [('Mass',f'{gp["mass_g"]:.0f} g'),('Friction μ',f'{gp["mu"]:.2f}'),
        ('Stiffness k',f'{gp["k"]:.0f} N/m'),('Aperture',f'{gp["aperture_mm"]:.1f} mm'),
        ('Initial force',f'{gp["initial_force"]:.3f} N'),
        ('Add force/retry',f'{gp["add_force"]:.3f} N'),
        ('Add closure/retry',f'{gp["closure_mm"]:.1f} mm')]
for k2, v in rows: print(f'  {k2:<20} {v}')

fig, (a1, a2) = plt.subplots(1, 2, figsize=(12, 4))
a1.bar(['Mass (g)', 'k (N/m)'], [gp['mass_g'], gp['k']], color=['steelblue','orange'])
a1.set_title('Object Properties')
a2.bar(['Aperture\n(mm)', 'Init force\n(N)', 'Add force\n(N)', 'Add closure\n(mm)'],
       [gp['aperture_mm'], gp['initial_force'], gp['add_force'], gp['closure_mm']],
       color=['tomato','purple','green','darkorange'])
a2.set_title('Gripper Parameters')
plt.suptitle(f'DeliGrasp Descriptor — "{ACTIVE_OBJECT}"', fontsize=12)
plt.tight_layout(); plt.show()
````

## 11. DeliGrasp — Adaptive Grasp
Close at `max(initial_force, 5N)`, read `/gripper/state`, retighten if measured force < threshold. Shows the full slip-detection loop.

````python
_SLIP_RATIO  = 0.4
CLOSE_FORCE  = max(gp['initial_force'], 5.0)
THRESHOLD    = max(0.15, gp['initial_force'] * _SLIP_RATIO)
MAX_RETRIES  = 3

print(f'initial_force = {gp["initial_force"]:.3f} N  |  CLOSE_FORCE = {CLOSE_FORCE:.1f} N')
print(f'contact threshold = {THRESHOLD:.3f} N  (= initial_force x {_SLIP_RATIO})')
print(f'max retries = {MAX_RETRIES}  |  max force cap = 16 N')
print()

node.disable_teach()
node.open_g(); time.sleep(0.5)

attempts = []
cur_force = CLOSE_FORCE

for attempt in range(MAX_RETRIES + 1):
    print(f'Attempt {attempt+1}: set_force={cur_force:.2f} N')
    node.set_force(cur_force)
    node.close_g()
    time.sleep(3.5)
    for _ in range(8): rclpy.spin_once(node, timeout_sec=0.15)
    s = node.gs

    if s is None:
        print('  [WARN] No gripper state')
        attempts.append({'a':attempt+1,'set':cur_force,'meas':0.0,'ap':0.0,'status':'no data'})
        break

    meas, ap = s.force, s.position
    status   = 'contact' if meas >= THRESHOLD else 'slip'
    print(f'  aperture={ap:.1f}mm  measured={meas:.3f}N  threshold={THRESHOLD:.3f}N  -> {status}')
    attempts.append({'a':attempt+1,'set':cur_force,'meas':meas,'ap':ap,'status':status})

    if status == 'contact':
        print(f'  Contact confirmed at {meas:.3f} N'); break
    if attempt == MAX_RETRIES:
        print('  Max retries reached — proceeding'); break
    cur_force = min(16.0, cur_force + gp['add_force'])

node.open_g()

# Plot
att  = [str(a['a']) for a in attempts]
meas = [a['meas'] for a in attempts]
aps  = [a['ap']   for a in attempts]
clrs = ['limegreen' if a['status']=='contact' else 'tomato' for a in attempts]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
ax1.bar(att, meas, color=clrs)
ax1.axhline(THRESHOLD, color='orange', ls='--', lw=2, label=f'threshold {THRESHOLD:.2f}N')
ax1.axhline(gp['initial_force'], color='purple', ls=':', lw=1.5, label=f'initial_force {gp["initial_force"]:.2f}N')
ax1.set_xlabel('Attempt'); ax1.set_ylabel('N'); ax1.set_title('Measured Force'); ax1.legend()
ax2.bar(att, aps, color='steelblue')
ax2.set_xlabel('Attempt'); ax2.set_ylabel('mm'); ax2.set_title('Gripper Aperture')
plt.suptitle(f'DeliGrasp Adaptive Grasp — "{ACTIVE_OBJECT}"  (green=contact, red=slip)', fontsize=12)
plt.tight_layout(); plt.show()

final = attempts[-1]
print(f'\nResult: {final["status"]}  |  aperture={final["ap"]:.1f}mm  force={final["meas"]:.3f}N  after {len(attempts)} attempt(s)')
````
